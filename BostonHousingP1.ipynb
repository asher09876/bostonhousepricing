```python
## CREATING DATA FRAME

df=pd.DataFrame(boston)
print(df.head())
print(df.info())
print(df.isnull())
print(df.isnull().sum())
```

```python
# independent and dependent features

X=df.iloc[:,:-1] #independent
y=df.iloc[:,-1] # dependent
```

TRAIN TEST SPLIT

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

# standardize the dataset
# aim is to come to global minimum
# for which all independent features should be in same scale


```python
#normalize
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
#linear regression
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
regression.fit(X_train,y_train)
```

```python
## print coefficients and intercepts
print(regression.coef_)
print(regression.intercept_)
```

```python
## on which parameters the model has been trained
regression.get_params()
```

```python
#### predict with test data
reg_pred=regression.predict(X_test)
print(reg_pred)
```

```python
## plot a scatter plot for the prediction
# if model is linear then its pridicting well
plt.scatter(y_test,reg_pred)
plt.xlabel("y_test")
plt.ylabel("regresiion prediction")
plt.show()
```

```python
# residual gives error
residual=y_test-reg_pred

# plot residual
sns.displot(residual,kind='kde')
plt.show()
```

```python
## scatter plot wrt to prediction and residuals
## distribution is uniform
plt.scatter(reg_pred,residual)
plt.xlabel("regresiion prediction")
plt.ylabel("residual")
plt.show()
```

from graph we can see model performs well

### testing performance

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
print("mean_absolute_error")
print(mean_absolute_error(y_test,reg_pred))
print("mean_squared_error")
print(np.sqrt(mean_squared_error(y_test,reg_pred))) # root mean sq


#r sq and adjusted r sq
# adjusted r sq always less than r sq
# the closer the value is to 1 its better

from sklearn.metrics import r2_score
score=r2_score(y_test,reg_pred)
print("r sq")
print(score)
print("adjusted mean sq")

## adjusted r sq
print(1- (1-score)*(len(y_test)-1) / (len(y_test) - X_test.shape[1] -1 ))
```
